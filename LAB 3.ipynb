```python
roll_no = 1024160062

L = [int(digit) * 10 for digit in str(roll_no)]

print("Initial L:", L)

L.append(50)
print("After append:", L)

L.insert(2, 70)
print("After insert:", L)

L.remove(20)
print("After remove:", L)

L.pop(3)
print("After pop:", L)

L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)

print("First three and last three:", L[:3], L[-3:])

average = sum(L) / len(L)

new_L = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", new_L)
```

```python
L = [10, 0, 20, 40, 10, 60, 0, 0, 60, 20]

scores = tuple(L[:8])
print("Scores:", scores)

highest = max(scores)
highest_index = scores.index(highest)

lowest = min(scores)
lowest_count = scores.count(lowest)

print("Highest score:", highest)
print("Index of highest:", highest_index)
print("Lowest score:", lowest)
print("Lowest appears:", lowest_count, "times")

reversed_scores = list(scores[::-1])
print("Reversed tuple as list:", reversed_scores)

x = int(input("Enter a score: "))

if x in scores:
    print("First occurrence index:", scores.index(x))
else:
    print("Score not present")

scores[0] = 100
```

```python
import random

random.seed(1024160062)

L = [random.randint(100, 900) for i in range(100)]

odd = [x for x in L if x % 2 != 0]
even = [x for x in L if x % 2 == 0]

print("Odd numbers:", odd)
print("Count of odd numbers:", len(odd))

print("Even numbers:", even)
print("Count of even numbers:", len(even))


def prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


primes = [x for x in L if prime(x)]

print("Prime numbers:", primes)
print("Count of prime numbers:", len(primes))


most_frequent = max(set(L), key=L.count)

print("Most frequent number:", most_frequent)
print("Number of occurrences:", L.count(most_frequent))
```
